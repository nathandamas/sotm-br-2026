# 10 — Substantive interpretation and spatial analysis of the OSM typology

This notebook performs **no API requests** and does not alter the typology defined in Notebook 09. It characterizes the three operational classes, formally tests differences between them, evaluates spatial coherence, and produces article-ready tables and figures.

The structural no-road class is interpreted separately where road-history variables are undefined. Road-present clusters are compared directly for growth, saturation, plateau, persistence, and contribution-dynamics indicators.

All inferential results are descriptive of the intrinsic OpenStreetMap history represented by the dataset. They are not tests of external completeness, positional accuracy, real-world road absence, or socioeconomic condition.


```python
from pathlib import Path
import runpy
from matplotlib.axes import Axes

# Compatibility shim for Matplotlib >= 3.11, where Axes.boxplot() renamed
# the `labels` keyword to `tick_labels`. The analysis script remains
# backward-compatible without changing any statistical logic.
_original_boxplot = Axes.boxplot
def _boxplot_compat(self, *args, **kwargs):
    if 'labels' in kwargs and 'tick_labels' not in kwargs:
        kwargs['tick_labels'] = kwargs.pop('labels')
    return _original_boxplot(self, *args, **kwargs)
Axes.boxplot = _boxplot_compat

script = Path('scripts/notebook10_interpret_osm_typology.py')
if not script.exists():
    script = Path('..') / 'scripts' / 'notebook10_interpret_osm_typology.py'

print('Executing:', script.resolve())
runpy.run_path(str(script), run_name='__main__')
```

## Outputs for manuscript development

The analysis exports class summaries, original-scale indicator profiles, Kruskal–Wallis tests, pairwise Mann–Whitney tests with Cliff's delta, focused road-present cluster comparisons, Fisher exact tests for selected binary states, Queen-neighbor spatial-coherence diagnostics, connected-component summaries, and manuscript-ready PNG/LaTeX outputs.

Cluster names remain provisional. The stored classification is not renamed automatically from statistical differences alone.
