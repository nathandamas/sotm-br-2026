# 12 — Article-ready choropleth maps

This notebook produces reproducible choropleth maps from the final 1 km OpenStreetMap typology dataset. Maps use the viridis palette, CARTO Positron when available, a geographic graticule, north arrow, graphic scale, CRS/SRC information, data-source statement, and explicit basemap attribution.

The map CRS is SIRGAS 2000 / UTM zone 23S (EPSG:31983), while the overlaid graticule is expressed in geographic longitude/latitude.


```python
from pathlib import Path
import runpy
script = Path('scripts/notebook12_build_article_maps.py')
if not script.exists(): script = Path('..') / script
print('Executing:', script.resolve())
runpy.run_path(str(script), run_name='__main__')
```

## Outputs

Each map is exported as high-resolution PNG and PDF under `figures/article/`. The final typology map uses three discrete viridis colors; continuous indicators use viridis with robust display limits. The contribution map uses logarithmic color normalization because of its strongly right-skewed distribution.
