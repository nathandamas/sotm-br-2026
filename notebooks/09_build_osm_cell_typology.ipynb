# 09 — Build and validate an intrinsic OSM cell typology

This notebook constructs a provisional typology for the 895 Petrópolis 1 km cells from the validated indicators and diagnostics produced by Notebooks 07–08. It performs **no API requests**.

The primary clustering uses a reduced, interpretable set of original indicators, not PCA scores alone. Cells with no current mapped road length are retained as an explicit structural class rather than receiving artificial road-history values through median imputation. For road-present cells, K-means and Ward hierarchical clustering are compared for k=2..7 using robust scaling, silhouette, Calinski–Harabasz, Davies–Bouldin, minimum cluster size, subsample stability (ARI), and Queen-neighbor agreement.

Cluster labels are deliberately provisional and non-semantic until their original-scale profiles are interpreted substantively.


```python
from pathlib import Path
import runpy
script = Path('scripts/notebook09_build_osm_cell_typology.py')
if not script.exists():
    script = Path('..') / 'scripts' / 'notebook09_build_osm_cell_typology.py'
print('Executing:', script.resolve())
runpy.run_path(str(script), run_name='__main__')
```

## Interpretation boundary

The resulting classes summarize **intrinsic OpenStreetMap mapping-history patterns**. They are not classes of external completeness, positional accuracy, road-system quality, urban development, or socioeconomic condition. Spatial coherence is reported as a diagnostic rather than imposed as a clustering constraint.
