# 14 — Query unique contributors for new article maps

This notebook adds two directly comparable contributor metrics for the 895 Petrópolis 1 km cells.

1. **All OSM unique contributors**: unique users who contributed to any OSM feature in each cell over the frozen Notebook 05 interval.
2. **Road-mapping unique contributors**: the already validated Notebook 05 count restricted to line features matching `highway=* and geometry:line`.

The all-OSM metric is queried with `/users/count/groupBy/boundary` without a feature filter. The road metric is not re-queried; it is reused from Notebook 05 so both maps remain temporally aligned with the existing analytical pipeline.


```python
from pathlib import Path
import runpy
script = Path('scripts/notebook14_query_unique_contributors.py')
if not script.exists():
    script = Path('..') / script
print('Executing:', script.resolve())
runpy.run_path(str(script), run_name='__main__')
```

## Outputs

The notebook writes `data/output/notebooks/14/petropolis_unique_contributors_895_cells.parquet`, plus quality-control and summary CSV files. The article-map workflow then merges these metrics into the final 1 km grid and generates Figures 09 and 10 with the same cartographic design as Figures 01–08.
